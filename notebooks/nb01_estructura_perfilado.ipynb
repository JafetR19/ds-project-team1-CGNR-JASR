# 5.1 Contexto y estructura
¿Qué fenómeno registra la base? ¿Quién la produjo y con qué propósito?

La base de datos contiene información de los incidentes que se reportan al C5 de la CDMX.
¿Cuántas filas y columnas tiene?
La base de datos tiene aproximadamente 124.3 mil filas y 14 columnas
¿Qué significa cada variable? Descríbanla con sus propias palabras, apoyándose en el diccionario de datos.

Columnas

**folio**: identificador que nos permite contabilizar todos los incidentes que se reportan aunque no necesariamente hagan referencia a incidentes distintos, ya que un incidente puede estar reportado más de una vez

**fecha_hora_creacion**: fecha exacta en la que se generó el reporte

**fecha_hora_cierre**: fecha en la que se concluyó la atención del reporte y se registró en el sistema

**tipo_incidente_c4**: Se reporta una categoría de accidente e.g. accidente, lesionado, "mi calle", "mi taxi", cadáver

**incidente_c4**: se agregan más detalles del incidente, e.g si hubo motociclistas, si fue un choque sin lesionados, etc.

**alcaldia_inicio**: alcaldía donde ocurrió el incidente

**codigo_cierre**: Es una clasificación que divide cada incidente en cuatro letras {A,F,D,I} y se refieren a A = “Afirmativo” F = “Falso” D = “Duplicados” I = “Informativo”

**clas_con_f_alarma**: Aqui se reportan las caterorias emergencia, unidades médicas, falsa alarma, delito, incidentes externos

**tipo_entrada**: Es el medio por el cuál se realiza el reporte del incidente

**alcaldia_cierre**: Es la alcaldía donde se da resolución al incidente

**alcaldia_catalogo**: Nombre de la alcaldía, aunque deberían ser 16 algunos registros vienen con na

**colonia_catalogo**: se incluyende los nombre de XXXXXXX colonias
longitud: coordenadas de la longitud

**latitud**: coordenadas de la latitud

# 5.2. Entidades y eventos 
**¿Cuál es la unidad de observación de cada fila?**
La unidad de análisis es cada reporte que se realiza

**¿Cuáles son las entidades relevantes y cómo se identifican?**

Depende del análisis que se quiera realizar, las entidades relevantes podrían ser aquello incidentes que se reportaron y que sí fueron afirmativos o los que se reportaron y luego resultaron falsos

**¿Qué eventos aparecen y qué representan respecto de esas entidades?**

Los eventos corresponden a las incidencias, es decir, nuestra unidad de análisis. Mientras que los eventos más relevantes de la entidad se encuentran en la variable de codigo_cierre donde podemos identificar si el incidente fue verdadero o no.

**¿Un folio identifica necesariamente un evento real distinto. Reflexionen sobre los duplicados**
Los folios no necesariamente identifican a un evento distinto ya que cada folio nuevo se genera cuando se realiza un reporte, el cual puede originarse de un evento repetido. La columna codigo_cierre nos ayudará a reconocer los folios que sí hacen referencia a un incidente distinto.

# 5.3 Tipos, calidad y limpieza
**Identifiquen tipos de datos y valores faltantes**
Todos los objetos se encuentran en formato str a excepción de latitud y longitud que son formato float 

**Revisen si las variables están representadas correctamente, especialmente fechas y horas**
No están bien representadas porque se encuentran en formato string.

**Hagan las conversiones necesarias y expliquen cada decisión**
Es necesario coercionar el tipo de valor a formato de fecha_hora_creacion y fecha_hora_cierre.

**Identifiquen posibles duplicados, categorías inconsistentes o coordenadas fuera del rango esperado, cuando aplique**
No identificamos valores duplicados en los folios. Sin embargo, los valores no son 100% consistentes dado que varían en el largo de la cadena de texto. Por ejemplo, algunos contienen 16 dígitos y otros 15 o 19. 

Asimismo, se detectaron 4 puntos fuera de la Ciudad de México.

**Coordenadas fuera del rango esperado**

De acuerdo con el INEGI las cordenadas correspondientes a la CDMX son Longitud99°21'53.64" W 98°56'25.08" W, Latitud19°02'53.52" N 19°35'34.08" N

Utilizando esta fórmula 

                         $$G+M/60 + S/3600$$

tenemos que los rangos que deberían estar dentro del polígono son Longitud: -99.3649 a -98.9403 Latitud: 19.0482 a 19.5928

Con esta información se pueden rectificar los datos que no corresponde a incidentes ocurridos en la CDMX. Por otro lado, la variable alcaldia_catalogo presenta valores faltantes, de acuerdo con https://datos.cdmx.gob.mx/dataset carpetas-de-investigacion-fgj-de-la-ciudad-de-mexico 

A partir de diciembre de 2023 se incluyeron las columnas "colonia_catalogo" y "alcaldia_catalogo". Estos campos se generaron a partir del cruce de la georreferenciación (latitud y longitud) y si analizamos los ausentes de dicha columa podemos ver que los faltantes estan presentes entre 2022 y 2024 por lo que se podrían completar con la información de la latitud y longitud https://en.www.inegi.org.mx/contenidos/app/areasgeograficas/resumen/resumen_09.pdf página 6
