# 5.1 Contexto y estructura
**¿Qué fenómeno registra la base? ¿Quién la produjo y con qué propósito?**
Fuente de los datos y objetivo del análisis: La fuente Incidentes viales reportados por C5 del gobierno de la CDMX, el objetivo del análisis es contar con una visión general del tipo de indicentes que se reportan, desde qué medios y alcaldías. **¿Cuántas filas y columnas tiene?**
La base de datos tiene aproximadamente 124.3 mil filas y 14 columnas
¿Qué significa cada variable? Descríbanla con sus propias palabras, apoyándose en el diccionario de datos.

**Columnas**

**folio**: identificador que nos permite contabilizar todos los incidentes que se reportan aunque no necesariamente hagan referencia a incidentes distintos, ya que un incidente puede estar reportado más de una vez

**fecha_hora_creacion**: fecha exacta en la que se generó el reporte

**fecha_hora_cierre**: fecha en la que se concluyó la atención del reporte y se registró en el sistema

**tipo_incidente_c4**: Se reporta una categoría de accidente e.g. accidente, lesionado, "mi calle", "mi taxi", cadáver

**incidente_c4**: se agregan más detalles del incidente, e.g si hubo motociclistas, si fue un choque sin lesionados, etc.

**alcaldia_inicio**: alcaldía donde ocurrió el incidente

**codigo_cierre**: Es una clasificación que divide cada incidente en cuatro letras {A,F,D,I} y se refieren a A = “Afirmativo” F = “Falso” D = “Duplicados” I = “Informativo”

**clas_con_f_alarma**: Aqui se reportan las caterorias emergencia, unidades médicas, falsa alarma, delito, incidentes externos

**tipo_entrada**: Es el medio por el cuál se realiza el reporte del incidente

**alcaldia_cierre**: Es la alcaldía donde se da resolución al incidente

**alcaldia_catalogo**: Nombre de la alcaldía, aunque deberían ser 16 algunos registros vienen con na

**colonia_catalogo**: se incluyende los nombre de colonias

**longitud**: coordenadas de la longitud

**latitud**: coordenadas de la latitud

# 5.4. Data Profiling o perfilado por tipo de variable 

**Categóricas**: categorías, frecuencias y valores faltantes. 

**Numéricas**: medidas descriptivas y valores extremos, si existen. 

**Temporales:** fechas mínima y máxima, cobertura y granularidad. 

**Interpretación de los resultados**

La base de datos contiene siete variables categóricas, con entre 4 y 16 categorías únicas.

**En tipo_incidente_c4**, predominan los accidentes (89.14%), seguidos de los reportes de lesionados (10.50%), mientras que las demás categorías tienen una participación marginal.

**En incidente_c4**, los choques sin lesionados (46.12%) y con lesionados (29.02%) concentran el 75.14% de los registros. Les siguen los atropellamientos (10.60%) y los incidentes relacionados con motociclistas (9.89%).

En general, se observa una alta concentración de los reportes en accidentes vehiculares, particularmente choques.

**En codigo_cierre**, predominan los registros clasificados como D, es decir duplicado, (42.57%), seguidos de A (36.54%) y F (20.22%). Por su parte, la categoría I (0.67%) presenta la menor frecuencia. Esto evidencia una proporción importante de reportes clasificados como duplicados.

**En clas_con_f_alarma**, predominan los reportes clasificados como Urgencias médicas (50.86%) y Emergencia (49.04%), que en conjunto concentran el 99.90% de los registros. Por su parte, las categorías Delito, Falsa alarma e Incidentes externos presentan una participación marginal.

**En tipo_entrada**, predominan los reportes recibidos mediante llamadas al 911 (89.47%), seguidos del botón de auxilio (5.70%) y radio (4.29%). Los demás medios de entrada presentan una participación marginal, lo que evidencia que las llamadas al 911 constituyen el principal canal de reporte de incidentes en la CDMX.

Para las numéricas el tiempo_minutos registra un promedio de 263.70 minutos, con una mediana de 187.63 minutos. La desviación estándar de 300.54 minutos y el valor máximo de 4,600.17 minutos sugieren la presencia de valores extremos y una distribución sesgada hacia la derecha. Esto indica que algunos incidentes presentan tiempos de cierre considerablemente superiores a los habituales.

**Las variables fecha_hora_creacion y fecha_hora_cierre** presentan una cobertura temporal de diciembre de 2021 a febrero de 2024, con registros a nivel de segundos y sin valores faltantes. Esto permite analizar la evolución de los incidentes y sus tiempos de atención a lo largo del periodo.

```python
### Creación de variables para más análisis 

df['hora'] = df['fecha_hora_creacion'].dt.hour
df['dia'] = df['fecha_hora_creacion'].dt.day
df['mes'] = df['fecha_hora_creacion'].dt.month
df['año'] = df['fecha_hora_creacion'].dt.year
df['tiempo_minutos'] = (df['fecha_hora_cierre'] - df['fecha_hora_creacion']).dt.total_seconds() / 60
```

```python
dias = { 0: 'Lunes',1: 'Martes',2: 'Miércoles',3: 'Jueves',4: 'Viernes',5: 'Sábado',6: 'Domingo'}
df['dia_semana'] = (df['fecha_hora_creacion'].dt.dayofweek.map(dias))
```

# **Identificar el número de categorías**

```python
def tabla_frecuencias(df, columna):
    tabla = (df[columna].value_counts().reset_index())
    tabla.columns = ['categoria', 'frecuencia']
    tabla['porcentaje'] = (
        100 * tabla['frecuencia'] / df.shape[0]).round(2)

    return tabla
```

# 3. Variables Temporales

Se identifica el periodo cubierto, el número de meses con registros y la distribución mensual.

```python
fechas = ['fecha_hora_creacion', 'fecha_hora_cierre']
df[fechas] = df[fechas].apply(pd.to_datetime, errors='coerce')

resumen_fechas = pd.DataFrame({
    'fecha_minima': df[fechas].min(),
    'fecha_maxima': df[fechas].max(),
    'faltantes': df[fechas].isna().sum()
})
display(resumen_fechas)
```
