# 5.1 Contexto y estructura
¿Qué fenómeno registra la base? ¿Quién la produjo y con qué propósito?
La base de datos contiene información de los incidentes que se reportan al C5 de la CDMX.
¿Cuántas filas y columnas tiene?
La base de datos tiene aproximadamente 124.3 mil filas y 14 columnas
¿Qué significa cada variable? Descríbanla con sus propias palabras, apoyándose en el diccionario de datos.

Columnas

folio: identificador que nos permite contabilizar todos los incidentes que se reportan aunque no necesariamente hagan referencia a incidentes distintos, ya que un incidente puede estar reportado más de una vez
fecha_hora_creacion: fecha exacta en la que se generó el reporte
fecha_hora_cierre: fecha en la que se concluyó la atención del reporte y se registró en el sistema
tipo_incidente_c4: Se reporta una categoría de accidente e.g. accidente, lesionado, "mi calle", "mi taxi", cadáver
incidente_c4: se agregan más detalles del incidente, e.g si hubo motociclistas, si fue un choque sin lesionados, etc.
alcaldia_inicio: alcaldía donde ocurrió el incidente
codigo_cierre: Es una clasificación que divide cada incidente en cuatro letras {A,F,D,I} y se refieren a A = “Afirmativo” F = “Falso” D = “Duplicados” I = “Informativo”
clas_con_f_alarma: Aqui se reportan las caterorias emergencia, unidades médicas, falsa alarma, delito, incidentes externos
tipo_entrada: Es el medio por el cuál se realiza el reporte del incidente
alcaldia_cierre: Es la alcaldía donde se da resolución al incidente
alcaldia_catalogo: Nombre de la alcaldía, aunque deberían ser 16 algunos registros vienen con na
colonia_catalogo: se incluyende los nombre de XXXXXXX colonias
longitud: coordenadas de la longitud
latitud: coordenadas de la latitud
